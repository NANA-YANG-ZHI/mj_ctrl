```python
# import mujoco
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# import numpy as np


def plot_contact_data(file_path):
    a = np.load(file_path)

    DT = 0.001
    n_plot = int(2.0 / DT)  # first 2s
    t = np.arange(n_plot) * DT

    skip = int(1.0 / DT)

    fe = a["force_error"][skip:]
    print(f"Mean |force error|: {np.mean(np.abs(fe)):.3f} N")
    print(f"Max  |force error|: {np.max(np.abs(fe)):.3f} N")

    fig, (ax_force, ax_vel) = plt.subplots(2, 1, sharex=True, figsize=(8, 5))

    ax_force.plot(t, a["force_error"][:n_plot], linewidth=0.9)
    ax_force.axhline(0, color="gray", linestyle="--", linewidth=0.7, alpha=0.6)
    ax_force.set_ylabel("Force Z Error (N)")

    vel_xy = np.linalg.norm(a["actual_velocitys"][:n_plot, :2], axis=1)
    ax_vel.plot(t, vel_xy, linewidth=0.9)
    ax_vel.set_ylabel("XY Velocity (m/s)")
    ax_vel.set_xlabel("Time (s)")

    plt.tight_layout()
    plt.show()
```

```python

csv_path = r"C:\Users\11452\wkspace\fork_mj_ctrl\sin_trajectory\wandb_export_2026-04-20T01_59_15.852+02_00.csv"
df = pd.read_csv(csv_path)

col = "eval_ppo_run_20260304_115524 - force_error"
DT = 0.02  # PPO step = 20 physics steps × 0.001s

fe = df[col].values
t  = df["Step"].values * DT

skip = int(1.0 / DT)  # skip first 1s = 50 steps
fe_ss = fe[skip:]
print(f"Mean |force error| (skip 1s): {np.mean(np.abs(fe_ss)):.3f} N")
print(f"Max  |force error| (skip 1s): {np.max(np.abs(fe_ss)):.3f} N")

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(t, fe, linewidth=0.9)
ax.axhline(0, color="gray", linestyle="--", linewidth=0.7, alpha=0.6)
ax.axvline(1.0, color="red", linestyle=":", linewidth=0.8, alpha=0.7, label="skip boundary (1s)")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Force Error (N)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
def print_model_info(model):
    """Print information about bodies and geoms in the model"""
    print("\n=== MODEL INFORMATION ===")
    print("Bodies:")
    for i in range(model.nbody):
        body_name = mujoco.mj_id2name(model, mujoco.mjtObj.mjOBJ_BODY, i)
        if body_name:
            print(f"  {i}: {body_name}")
    
    print("\nGeoms:")
    for i in range(model.ngeom):
        geom_name = mujoco.mj_id2name(model, mujoco.mjtObj.mjOBJ_GEOM, i)
        if geom_name:
            body_id = model.geom_bodyid[i]
            body_name = mujoco.mj_id2name(model, mujoco.mjtObj.mjOBJ_BODY, body_id)
            print(f"  {i}: {geom_name} (body: {body_name})")
```

```python
def get_table_position(model, data):
    """Get the table's position in world coordinates"""
    try:
        # Try to find table body
        table_body_id = mujoco.mj_name2id(model, mujoco.mjtObj.mjOBJ_BODY, "table")
        table_pos = data.xpos[table_body_id].copy()
        
        # Table surface is typically higher than body center
        table_height = table_pos[2] + 0.4  # Estimate table surface height
        table_surface_pos = np.array([table_pos[0], table_pos[1], table_height])
        
        return table_surface_pos
    except:
        # Fallback: use position from XML (table at 0.75, 0, 0 with ~0.4m height)
        # return np.array([0.75, 0.0, 0.4])
        return 0

mujoco.mj_forward(model, data)    
get_table_position(model, data)
```
